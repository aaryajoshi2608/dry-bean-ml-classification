```python

#  Load and Display Dataset

import pandas as pd
import numpy as np

df = pd.read_csv("Dry_Beans_Dataset.csv")

print("Dataset Shape:", df.shape)

print("\nFirst 5 Rows:")
display(df.head())
```

```python

#  Data Exploration and Quality Checks

print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nNumber of Duplicate Rows:")
print(df.duplicated().sum())

print("\nClass Distribution:")
print(df["Class"].value_counts())

print("\nStatistical Summary:")
display(df.describe())
```

```python

#  Remove Duplicate Records

print("Original Dataset Shape:", df.shape)

df = df.drop_duplicates().reset_index(drop=True)

print("Dataset Shape After Removing Duplicates:", df.shape)

print("Remaining Duplicate Rows:", df.duplicated().sum())
```

```python

#  Separate Features and Target Variable

X = df.drop("Class", axis=1)

y = df["Class"]

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)

print("\nTarget Classes:")
print(y.unique())
```

```python

#  Split Dataset into Training and Testing Sets

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Features Shape:", X_train.shape)
print("Testing Features Shape:", X_test.shape)

print("Training Target Shape:", y_train.shape)
print("Testing Target Shape:", y_test.shape)
```

```python

#  Feature Scaling

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

print("Training Data Shape:", X_train_scaled.shape)
print("Testing Data Shape:", X_test_scaled.shape)

print("\nFirst 5 Rows of Scaled Training Data:")
print(X_train_scaled[:5])
```

```python

#  Logistic Regression Model Training

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

lr_model = LogisticRegression(max_iter=2000, random_state=42)

lr_model.fit(X_train_scaled, y_train)

y_pred_lr = lr_model.predict(X_test_scaled)

lr_accuracy = accuracy_score(y_test, y_pred_lr)

print("Logistic Regression Model Trained Successfully!")
print("Accuracy:", round(lr_accuracy, 4))
```

```python

#  Logistic Regression Evaluation Metrics

from sklearn.metrics import accuracy_score, roc_auc_score
from sklearn.metrics import precision_score, recall_score
from sklearn.metrics import f1_score, matthews_corrcoef

y_prob_lr = lr_model.predict_proba(X_test_scaled)

lr_accuracy = accuracy_score(y_test, y_pred_lr)
lr_auc = roc_auc_score(y_test, y_prob_lr, multi_class="ovr", average="weighted")
lr_precision = precision_score(y_test, y_pred_lr, average="weighted", zero_division=0)
lr_recall = recall_score(y_test, y_pred_lr, average="weighted", zero_division=0)
lr_f1 = f1_score(y_test, y_pred_lr, average="weighted", zero_division=0)
lr_mcc = matthews_corrcoef(y_test, y_pred_lr)

print("Logistic Regression Evaluation Metrics")
print("Accuracy :", round(lr_accuracy, 4))
print("AUC      :", round(lr_auc, 4))
print("Precision:", round(lr_precision, 4))
print("Recall   :", round(lr_recall, 4))
print("F1 Score :", round(lr_f1, 4))
print("MCC Score:", round(lr_mcc, 4))
```

```python

#  Decision Tree Classifier

from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

dt_model = DecisionTreeClassifier(random_state=42)

dt_model.fit(X_train, y_train)

y_pred_dt = dt_model.predict(X_test)

dt_accuracy = accuracy_score(y_test, y_pred_dt)

print("Decision Tree Model Trained Successfully!")
print("Accuracy:", round(dt_accuracy, 4))
```

```python

#  Decision Tree Evaluation Metrics

y_prob_dt = dt_model.predict_proba(X_test)

dt_accuracy = accuracy_score(y_test, y_pred_dt)
dt_auc = roc_auc_score(y_test, y_prob_dt, multi_class="ovr", average="weighted")
dt_precision = precision_score(y_test, y_pred_dt, average="weighted", zero_division=0)
dt_recall = recall_score(y_test, y_pred_dt, average="weighted", zero_division=0)
dt_f1 = f1_score(y_test, y_pred_dt, average="weighted", zero_division=0)
dt_mcc = matthews_corrcoef(y_test, y_pred_dt)

print("Decision Tree Evaluation Metrics")
print("Accuracy :", round(dt_accuracy, 4))
print("AUC      :", round(dt_auc, 4))
print("Precision:", round(dt_precision, 4))
print("Recall   :", round(dt_recall, 4))
print("F1 Score :", round(dt_f1, 4))
print("MCC Score:", round(dt_mcc, 4))
```

```python

#  K-Nearest Neighbors Classifier

from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

knn_model = KNeighborsClassifier(n_neighbors=5)

knn_model.fit(X_train_scaled, y_train)

y_pred_knn = knn_model.predict(X_test_scaled)

knn_accuracy = accuracy_score(y_test, y_pred_knn)

print("KNN Model Trained Successfully!")
print("Accuracy:", round(knn_accuracy, 4))
```

```python

#  KNN Evaluation Metrics

y_prob_knn = knn_model.predict_proba(X_test_scaled)

knn_accuracy = accuracy_score(y_test, y_pred_knn)
knn_auc = roc_auc_score(y_test, y_prob_knn, multi_class="ovr", average="weighted")
knn_precision = precision_score(y_test, y_pred_knn, average="weighted", zero_division=0)
knn_recall = recall_score(y_test, y_pred_knn, average="weighted", zero_division=0)
knn_f1 = f1_score(y_test, y_pred_knn, average="weighted", zero_division=0)
knn_mcc = matthews_corrcoef(y_test, y_pred_knn)

print("KNN Evaluation Metrics")
print("Accuracy :", round(knn_accuracy, 4))
print("AUC      :", round(knn_auc, 4))
print("Precision:", round(knn_precision, 4))
print("Recall   :", round(knn_recall, 4))
print("F1 Score :", round(knn_f1, 4))
print("MCC Score:", round(knn_mcc, 4))
```

```python

# Gaussian Naive Bayes Classifier

from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

nb_model = GaussianNB()

nb_model.fit(X_train_scaled, y_train)

y_pred_nb = nb_model.predict(X_test_scaled)

nb_accuracy = accuracy_score(y_test, y_pred_nb)

print("Gaussian Naive Bayes Model Trained Successfully!")
print("Accuracy:", round(nb_accuracy, 4))
```

```python

#  Gaussian Naive Bayes Evaluation Metrics

y_prob_nb = nb_model.predict_proba(X_test_scaled)

nb_accuracy = accuracy_score(y_test, y_pred_nb)
nb_auc = roc_auc_score(y_test, y_prob_nb, multi_class="ovr", average="weighted")
nb_precision = precision_score(y_test, y_pred_nb, average="weighted", zero_division=0)
nb_recall = recall_score(y_test, y_pred_nb, average="weighted", zero_division=0)
nb_f1 = f1_score(y_test, y_pred_nb, average="weighted", zero_division=0)
nb_mcc = matthews_corrcoef(y_test, y_pred_nb)

print("Gaussian Naive Bayes Evaluation Metrics")
print("Accuracy :", round(nb_accuracy, 4))
print("AUC      :", round(nb_auc, 4))
print("Precision:", round(nb_precision, 4))
print("Recall   :", round(nb_recall, 4))
print("F1 Score :", round(nb_f1, 4))
print("MCC Score:", round(nb_mcc, 4))
```

```python

#  Support Vector Machine Classifier

from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

svm_model = SVC(kernel="rbf", probability=True, random_state=42)

svm_model.fit(X_train_scaled, y_train)

y_pred_svm = svm_model.predict(X_test_scaled)

svm_accuracy = accuracy_score(y_test, y_pred_svm)

print("SVM Model Trained Successfully!")
print("Accuracy:", round(svm_accuracy, 4))
```

```python

# SVM Evaluation Metrics

y_prob_svm = svm_model.predict_proba(X_test_scaled)

svm_accuracy = accuracy_score(y_test, y_pred_svm)
svm_auc = roc_auc_score(y_test, y_prob_svm, multi_class="ovr", average="weighted")
svm_precision = precision_score(y_test, y_pred_svm, average="weighted", zero_division=0)
svm_recall = recall_score(y_test, y_pred_svm, average="weighted", zero_division=0)
svm_f1 = f1_score(y_test, y_pred_svm, average="weighted", zero_division=0)
svm_mcc = matthews_corrcoef(y_test, y_pred_svm)

print("Support Vector Machine Evaluation Metrics")
print("Accuracy :", round(svm_accuracy, 4))
print("AUC      :", round(svm_auc, 4))
print("Precision:", round(svm_precision, 4))
print("Recall   :", round(svm_recall, 4))
print("F1 Score :", round(svm_f1, 4))
print("MCC Score:", round(svm_mcc, 4))
```

```python

# Comparison of All Five Classification Models

import pandas as pd

comparison_df = pd.DataFrame({
    "ML Model Name": [
        "Logistic Regression",
        "Decision Tree",
        "KNN",
        "Naive Bayes",
        "Support Vector Machine"
    ],
    "Accuracy": [
        lr_accuracy, dt_accuracy, knn_accuracy, nb_accuracy, svm_accuracy
    ],
    "AUC": [
        lr_auc, dt_auc, knn_auc, nb_auc, svm_auc
    ],
    "Precision": [
        lr_precision, dt_precision, knn_precision, nb_precision, svm_precision
    ],
    "Recall": [
        lr_recall, dt_recall, knn_recall, nb_recall, svm_recall
    ],
    "F1": [
        lr_f1, dt_f1, knn_f1, nb_f1, svm_f1
    ],
    "MCC": [
        lr_mcc, dt_mcc, knn_mcc, nb_mcc, svm_mcc
    ]
})

print("Comparison of All Five ML Models:")
display(comparison_df.round(4))

print("\nBest Model Based on Accuracy:")
print(comparison_df.loc[comparison_df["Accuracy"].idxmax(), "ML Model Name"])
```

```python

#  Model Performance Analysis

print("Model Performance Ranking Based on Accuracy:\n")

ranking_df = comparison_df.sort_values(
    by="Accuracy",
    ascending=False
).reset_index(drop=True)

display(ranking_df.round(4))

print("\nBest Performing Model:")
print(ranking_df.loc[0, "ML Model Name"])

print("\nHighest Accuracy:", round(ranking_df.loc[0, "Accuracy"], 4))

print("\nLowest Performing Model:")
print(ranking_df.loc[len(ranking_df)-1, "ML Model Name"])
```
